```python
import pandas as pd
import numpy as np
import sklearn
import matplotlib
import pickle
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('Churn_Modelling.csv')
df.head()
```

```python
df.drop(['RowNumber','CustomerId','Surname'],axis=1,inplace=True)
```

```python
df.head()
```

```python
label_encoder_gender = LabelEncoder()
df['Gender'] = label_encoder_gender.fit_transform(df['Gender'])
df.head()
```

```python
one_hot_encoder_geo = OneHotEncoder()
onehot_encoder_obj = one_hot_encoder_geo.fit_transform(df[['Geography']])
onehot_encoder_obj.toarray()
```

```python

df1 = pd.DataFrame(onehot_encoder_obj.toarray(),columns=one_hot_encoder_geo.get_feature_names_out())
df1

df = pd.concat([df.drop('Geography',axis=1),df1],axis=1)
df
```

```python
df
```

```python
X = df.drop('Exited',axis=1)
y = df[['Exited']]
X
```

```python
#now train test split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42,)
X_train
```

```python
scaler = StandardScaler()
X_train_standard = scaler.fit_transform(X=X_train)
X_test_standard = scaler.transform(X=X_test)
X_test_standard
```

```python
#now creating the picke files for the encoders and scalers for the features use
with open('label_encoder_gender.pkl','wb') as file:
    pickle.dump(label_encoder_gender,file)

with open('one_hot_encoder_geograph.pkl','wb') as file:
    pickle.dump(one_hot_encoder_geo,file)

with open('standard_scaler.pkl','wb') as file:
    pickle.dump(scaler,file)
```

```python
X_test_standard
```

```python
## ann implimentation
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
X_train_standard.shape[1]
```

```python
model = Sequential(
    [Dense(activation='relu',units=64,input_shape=(X_train_standard.shape[1],)), #hd1 connected with input layer 2
     Dense(activation='relu',units=32), #hd2 connected with hidden layer 1
     Dense(activation='sigmoid',units=1) # this will be the output layer connected with hidden layer 2
    ]
)

model.summary()
```

```python
import tensorflow
optimizer = tensorflow.keras.optimizers.Adam(learning_rate = 0.01)
loss_function = tensorflow.keras.losses.BinaryCrossentropy()

model.compile(loss=loss_function,optimizer=optimizer,metrics=['accuracy'])
```

```python
##setting up the tensor board
log_file_path = "log/fit/"+ datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorflow_callback = TensorBoard(log_dir=log_file_path,histogram_freq=1)
```

```python
## settting up the early stopp
early_stop_callback = EarlyStopping(monitor="val_loss",patience=5,restore_best_weights=True)
```

```python
history = model.fit(X_train_standard,y_train,validation_data=[X_test_standard,y_test],epochs=100,callbacks=[tensorflow_callback,early_stop_callback])
```

```python
model.save('model.h5')
```

```python
##now predicting the values based on input
input_data = {
    'CreditScore': 600,
    'Gender': 'Male',          # raw text if you also label-encode it
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000.0,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000.0,
    'Geography': 'France'      # one of 'France', 'Germany', 'Spain'
}
#loading the objects
with open('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender = pickle.load(file=file)
with open('one_hot_encoder_geograph.pkl','rb') as file:
    one_hot_encoder_geograph = pickle.load(file=file)
with open('standard_scaler.pkl','rb') as file:
    standard_scaler = pickle.load(file=file)

input_df = pd.DataFrame([input_data])
# now transform the data
geography_transformed_data = one_hot_encoder_geo.transform(input_df[['Geography']])
geography_transformed_data.toarray()

feature_names_geo = one_hot_encoder_geo.get_feature_names_out()

geo_df = pd.DataFrame(geography_transformed_data.toarray(),columns=feature_names_geo)
geo_df

#now concatenate geo_df to input_df dropping geography
new_input_df = pd.concat([input_df.drop('Geography',axis=1),geo_df],axis=1)
new_input_df

#now change the gender
new_input_df['Gender'] = label_encoder_gender.transform(new_input_df[['Gender']])[0]

#now standarizing the data
new_input_standard = standard_scaler.transform(new_input_df)
new_input_standard

## now predict the output with new input standard data

y_pred = model.predict(new_input_standard)
prediction_prob = y_pred[0][0]

if prediction_prob > 0.5 :
    print('The customer will be churned')
else :
    print('The customer will not be churned')
```
